(01) 최악의 경우: O(n), 최선의 경우: O(log n)

(02) AVL 트리는 삽입 또는 삭제 직후에 균형이 깨질 수 있다. 삽입의 경우 한쪽 서브트리의 높이가 증가하면서 불균형이 발생할 수 있고, 삭제의 경우 반대쪽 서브트리의 높이가 감소하면서 불균형이 발생할 수 있다. 따라서 주어진 트리는 삽입과 삭제 모두의 경우에 의해 발생할 수 있다.

(03) AVL 트리의 불균형은 삽입뿐 아니라 삭제 과정에서도 발생할 수 있다. 삽입 시 특정 방향으로 노드가 추가되면서 불균형이 발생할 수 있고, 삭제 시 한쪽 서브트리의 높이가 감소하면서 상대적으로 다른 쪽이 깊어져 불균형이 발생할 수 있다. 따라서 주어진 경우는 삽입과 삭제 모두 가능하다.

(04)
초기 트리:
```text
      40
     /  |
   10    55
  /  |
 5    20
     /
    17
```

이 트리는 40의 왼쪽 서브트리가 더 깊고, 그 왼쪽 자식 10의 오른쪽이 더 깊으므로 LR 타입이다.

1단계: 10을 기준으로 좌회전
```text
      40
     /  |
   20    55
  /
 10
 / |
5   17
```

2단계: 40을 기준으로 우회전
```text
      20
     /  |
   10    40
  / |      |
 5   17     55
 ```

따라서 최종 AVL 트리는 위와 같다.

(05) 두 번 회전 필요한 경우: LR(좌 -> 우), RL(우 -> 좌)

(06) X / 회전은 균형만 맞춤. 높이는 유지될 수도 있음.

(07) X / AVL은 최초 불균형 노드만 고치면 해결됨. 그 위는 자동으로 균형 유지됨.

(08)
초기 트리:
```text
        10(B)
       /     |
    5(R)     13(B)
   /   |     /   |
 3(B) 6(B) 11(R) 15(R)
 / |
2(R)4(R)
```
Step 1: 7 삽입
BST 규칙 따라 삽입 위치:
10 → 5 → 6 → 오른쪽
```text
        10(B)
       /     |
    5(R)     13(B)
   /   |     /   |
 3(B) 6(B) 11(R) 15(R)
 / |     |
2(R)4(R)  7(R)
```
새 노드 = Red

Step 2: 문제 발생
부모 = 6(B)
부모가 Black이면 문제 없음 (RB 조건 유지)

최종결과:
추가 수정 없음
```text
        10(B)
       /     |
     5(R)     13(B)
    /   |     /   |
  3(B) 6(B) 11(R) 15(R)
  / |     |
2(R)4(R)  7(R)
```
키 7을 삽입하면 BST 규칙에 따라 6의 오른쪽 자식으로 삽입된다.
새로 삽입된 노드는 Red로 설정되며, 부모 노드 6은 Black이므로 Red-Black 트리의 성질이 깨지지 않는다.

따라서 추가적인 회전이나 색 변경 없이 위와 같은 트리가 최종 결과가 된다.

(12)
초기 트리:
```text
                [15]
          /              |
       [3 8]            [19 22]
     /   |   |          /   |   |
 [1 2] [5 6] [9 10] [16 18] [20 21] [24 25 26]
```
22의 직후 원소는 오른쪽 자식 리프의 가장 작은 값인 24.

22를 24로 바꾸고, 리프에서 24 삭제:
```text
                [15]
          /              |
       [3 8]            [19 24]
     /   |   |          /   |    |
 [1 2] [5 6] [9 10] [16 18] [20 21] [25 26]
```
[25 26]은 키가 2개라서 underflow가 발생하지 않는다. 삭제 후 리프는 [25 26]이 되고 최소 키 수를 만족하므로 추가적인 재분배나 병합은 필요 없다.

(13)
12번 결과:
```text
                [15]
          /              |
       [3 8]            [19 24]
     /   |   |          /   |    |
 [1 2] [5 6] [9 10] [16 18] [20 21] [25 26]
```
19는 내부 노드에 있으므로, 19의 직후 원소(successor)인 20과 바꾼다.
```text
                [15]
          /              |
       [3 8]            [20 24]
     /   |   |          /   |    |
 [1 2] [5 6] [9 10] [16 18] [19 21] [25 26]
```
이제 리프 노드 [19 21]에서 19를 삭제한다.

최종 결과:
```text
                [15]
          /              |
       [3 8]            [20 24]
     /   |   |          /   |    |
 [1 2] [5 6] [9 10] [16 18] [21] [25 26]
```
[21]은 최소 키 수보다 적어서 underflow가 난다고 볼 수 있어. 형제에게서 빌리거나 병합해야 한다. 오른쪽 형제 [25 26]도 최소 키 수라 빌릴 수 없으므로, 부모의 24와 함께 병합한다.

최종 B-트리:
```text
                [15]
          /              |
       [3 8]             [20]
     /   |   |          /    |
 [1 2] [5 6] [9 10] [16 18] [21 24 25 26]
```
19는 내부 노드에 있으므로 직후 원소인 20과 교환한 뒤, 리프 노드에서 19를 삭제한다. 삭제 후 리프가 [21]이 되어 underflow가 발생한다. 오른쪽 형제 [25 26]에서 키를 빌릴 수 없으므로 부모의 24를 내려 병합한다. 따라서 최종 결과는 위와 같다.

(14) B-트리에서 리프가 아닌 노드에 있는 원소의 직후 원소(successor)는 그 원소의 오른쪽 자식 서브트리에서 가장 작은 값이다. 가장 작은 값을 찾으려면 오른쪽 서브트리로 한 번 내려간 뒤, 계속 가장 왼쪽 자식을 따라 내려간다. 더 이상 내려갈 수 없는 곳이 가장 작은 값이고, 그 위치는 항상 리프 노드이다. 따라서 리프가 아닌 노드에 있는 원소의 직후 원소는 항상 리프 노드에 있다.

(15) 한 키는 키 값 20바이트와 페이지 번호 4바이트로 구성되므로, 하나의 키에 필요한 공간은 24바이트이다. 디스크 블록의 크기가 8,192바이트이므로, 한 노드에 저장할 수 있는 최대 키 수는 다음과 같이 계산된다. 8192 ÷ 24 ≈ 341 / 따라서 한 노드는 최대 341개의 키를 가질 수 있도록 설계할 수 있다.

(16) B-트리에서 각 노드가 최대 k개의 키를 가질 수 있다면, 내부 노드는 최대 k+1개의 자식을 가진다. 리프 노드는 실제 데이터(키)를 저장하는 노드이므로 가능한 한 많은 키를 포함하게 되고, 내부 노드는 자식 노드들을 연결하는 역할을 하므로 리프 노드보다 상대적으로 적은 수의 키를 가진다. 따라서 전체적으로 보면, 리프 노드에 저장된 키의 수가 내부 노드에 저장된 키의 수보다 훨씬 많다.

(17) B-트리에서 언더플로우가 발생하여 형제 노드와 병합할 때, 두 노드의 키들과 부모 노드의 키 1개를 함께 합친다. 각 노드는 최대 k개의 키를 가질 수 있으며, 언더플로우가 발생한 노드와 그 형제 노드는 각각 최소 ⌈k/2⌉개의 키를 가지고 있다. 따라서 병합 후의 키 개수는 다음과 같다. ⌈k/2⌉ + ⌈k/2⌉ + 1 ≤ k / 즉, 병합 후에도 최대 키 수 k를 초과하지 않으므로 오버플로우는 발생하지 않는다.

(18)
각 노드가 최대 1,000개의 키를 가지는 B-트리에서, 자식 노드의 최대 개수는 1,001개이다.
따라서 트리의 높이 h일 때 저장할 수 있는 최대 키 수는 대략 다음과 같다.

```text
최대: (1001^h - 1)
최소: (501^h - 1)   (각 노드가 최소 ⌈1000/2⌉ = 500개 이상의 키를 가지므로)
```
주어진 키의 수는 1조 개(10^12)이므로,
최대한 꽉 찬 경우 (최소 높이)
```text
1001^h ≈ 10^12
→ h ≈ 4
```
최소한만 채운 경우 (최대 높이)
```text
501^h ≈ 10^12
→ h ≈ 5
```
B-트리의 높이는 약 4 ~ 5 사이이다.
높이 범위: 4 ≤ h ≤ 5

(19)
B-트리 인덱스는 디스크 접근 횟수를 줄이는 것이 중요하다.
트리의 모양을 바꾸지 않으면서 접근 횟수를 줄이려면, 자주 접근하는 노드를 메인 메모리에 캐싱하면 된다.

특히 루트 노드나 상위 레벨의 내부 노드는 거의 모든 검색에서 공통으로 접근하므로, 이 노드들을 메모리에 미리 올려두면 디스크 접근 횟수를 크게 줄일 수 있다.
```text
디스크에 저장된 B-트리
→ 루트 노드와 상위 내부 노드만 메모리에 캐싱
→ 검색 시 하위 노드에 접근할 때만 디스크 접근
```
따라서 실용적인 방법은 루트 노드와 자주 쓰이는 상위 노드를 메모리에 캐싱하는 것이다.
